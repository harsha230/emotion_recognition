## Family-Level Comparison

```python
import pandas as pd
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# 1. Define a function to map each model into its family
# ---------------------------------------------------------
def identify_family(model_name: str) -> str:
    model_name = model_name.lower()
    if "qwen" in model_name:
        return "Qwen3-VL"
    elif "llava" in model_name:
        return "LLaVA"
    elif "minicpm" in model_name:
        return "minicpm-v"
    elif "deepface" in model_name:
        return "deepface"
    else:
        return "Other"

# ---------------------------------------------------------
# 2. Create a new column "family"
# ---------------------------------------------------------
df["family"] = df["model_name"].apply(identify_family)

# ---------------------------------------------------------
# 3. Compute summary metrics by family
# ---------------------------------------------------------
metrics = ["accuracy", "macro_f1"]

family_summary = (
    df.groupby("family")[metrics]
    .mean()
    .sort_values(by="macro_f1", ascending=False)
)

print("Family-Level Performance Summary:")
print(family_summary)

# ---------------------------------------------------------
# 4. Plot grouped bar chart comparing families
# ---------------------------------------------------------
plt.figure(figsize=(12, 6))

family_summary.plot(kind="bar", figsize=(12, 6))
plt.title("Performance Comparison Across Model Families")
plt.xlabel("Model Family")
plt.ylabel("Score")
plt.xticks(rotation=0)
plt.legend(title="Metric")
plt.tight_layout()
plt.show()
```

## Comparison of Models with Similar Sizes

```python
import pandas as pd
import matplotlib.pyplot as plt
import re

# ---------------------------------------------------------
# 1. Extract approximate model size from the model name
# ---------------------------------------------------------
def extract_model_size(model_name: str) -> str:
    model_name = model_name.lower()

    # Look for patterns like "2b", "3.8b", "7b", "13b", "30b", "32b"
    match = re.search(r"(\d+\.?\d*)b", model_name)
    if match:
        return match.group(1) + "B"   # e.g., "4b" -> "4B"
    return "Unknown"

df["model_size"] = df["model_name"].apply(extract_model_size)

# ---------------------------------------------------------
# 2. Filter sizes that appear in more than one model
#    (i.e., sizes that allow comparison)
# ---------------------------------------------------------
size_counts = df["model_size"].value_counts()
comparable_sizes = size_counts[size_counts > 1].index.tolist()

df_comparable = df[df["model_size"].isin(comparable_sizes)].copy()

print("Comparable Model Sizes:")
print(size_counts[size_counts > 1])

# ---------------------------------------------------------
# 3. Select metrics for comparison
# ---------------------------------------------------------
metrics = ["macro_f1", "accuracy", "macro_precision", "macro_recall"]

# Create summary table
comparison_table = (
    df_comparable[["model_size", "model_name"] + metrics]
    .sort_values(by=["model_size", "macro_f1"], ascending=[True, False])
)

print("\nComparison Between Models of Similar Size:")
print(comparison_table)

# ---------------------------------------------------------
# 4. Plot grouped bar chart for macro-F1 by size
# ---------------------------------------------------------
plt.figure(figsize=(12, 6))

for size in comparable_sizes:
    subset = df_comparable[df_comparable["model_size"] == size]
    plt.bar(subset["model_name"], subset["macro_f1"], label=size)

plt.title("Macro-F1 Comparison Between Models of Similar Size")
plt.xlabel("Model Name")
plt.ylabel("Macro F1 Score")
plt.xticks(rotation=45, ha="right")
plt.legend(title="Model Size")
plt.tight_layout()
plt.show()
```

```python
# ---------------------------------------------------------
# Radar Chart: Family-Level Mean F1 Scores per Emotion Class
# ---------------------------------------------------------
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

emotion_cols = [col for col in df.columns if col.startswith("f1_")]

# Compute family averages
family_means = df.groupby("family")[emotion_cols].mean()

# Radar setup
labels = emotion_cols
num_vars = len(labels)
angles = np.linspace(0, 2 * np.pi, num_vars, endpoint=False).tolist()

# Plot each family
plt.figure(figsize=(10, 10))
for family in family_means.index:
    values = family_means.loc[family].tolist()
    values += values[:1]  # close the circle
    ax = plt.subplot(111, polar=True)
    ax.plot(angles + angles[:1], values, label=family)
    ax.fill(angles + angles[:1], values, alpha=0.15)

plt.title("Family-Level Mean F1 Scores by Emotion Class")
plt.legend(loc="upper right", bbox_to_anchor=(1.2, 1.1))
plt.tight_layout()
plt.show()
```

```python
# ---------------------------------------------------------
# Plot Confusion Matrix Heatmap for a Selected Model
# ---------------------------------------------------------
import pandas as pd
import seaborn as sns
import json
import matplotlib.pyplot as plt

def plot_confusion_matrix(model_name: str):
    row = df[df["model_name"] == model_name].iloc[0]

    # Load JSON confusion matrix
    matrix = json.loads(row["confusion_matrix_json"])

    # Convert to DataFrame
    cm_df = pd.DataFrame(matrix["matrix"], 
                         index=matrix["labels_true"], 
                         columns=matrix["labels_pred"])

    # Plot heatmap
    plt.figure(figsize=(10, 6))
    sns.heatmap(cm_df, annot=True, cmap="mako", fmt="d")
    plt.title(f"Confusion Matrix – {model_name}")
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.tight_layout()
    plt.show()

# Example usage:
# plot_confusion_matrix("qwen3-vl:8b-instruct")
```

```python
# ---------------------------------------------------------
# Summarize Family-Level Confusion Errors (Corrected Version)
# ---------------------------------------------------------
import pandas as pd
import json

def extract_confusion_errors(row):
    """
    Computes total misclassifications for confusion matrices structured as:
    { true_label: {pred_label: count, ...}, ... }
    """
    cm = json.loads(row["confusion_matrix_json"])

    total_off_diagonal = 0
    
    for true_label, preds in cm.items():
        for pred_label, count in preds.items():
            # Only count errors where prediction != true label
            if true_label != pred_label:
                total_off_diagonal += count

    return total_off_diagonal

# Apply function
df["total_confusion_errors"] = df.apply(extract_confusion_errors, axis=1)

# Summarize by family
family_confusion_summary = df.groupby("family")["total_confusion_errors"].mean()

print("Mean Cross-Class Confusion per Family:")
print(family_confusion_summary)
```

```python
# ---------------------------------------------------------
# Robust Recursive Invalid Count Extraction
# ---------------------------------------------------------
import pandas as pd
import json

def sum_numeric_values(obj):
    """
    Recursively sum all numeric values inside a JSON-like dict structure.
    """
    total = 0
    
    if isinstance(obj, dict):
        for value in obj.values():
            total += sum_numeric_values(value)
    elif isinstance(obj, list):
        for element in obj:
            total += sum_numeric_values(element)
    elif isinstance(obj, (int, float)):
        total += obj
    
    return total


def count_invalid(row):
    invalid_json = json.loads(row["invalid_counts_json"])
    return sum_numeric_values(invalid_json)


# Apply to dataframe
df["invalid_total"] = df.apply(count_invalid, axis=1)

print("Invalid Outputs per Model:")
print(df[["model_name", "invalid_total"]])

# Summary by family
family_invalid = df.groupby("family")["invalid_total"].mean()
print("\nInvalid Output Mean per Family:")
print(family_invalid)

# Plot
family_invalid.plot(kind="bar", figsize=(8, 4), title="Average Invalid Outputs per Family")
plt.ylabel("Invalid Count")
plt.tight_layout()
plt.show()
```
